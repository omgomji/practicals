## Aim
To perform basic word-level analysis on natural language text using tokenization, normalization, stop-word removal, frequency analysis, stemming, lemmatization, and part-of-speech tagging.

```python
import nltk
from collections import Counter

from nltk import pos_tag, word_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer, WordNetLemmatizer

resources = [
    "punkt",
    "punkt_tab",
    "stopwords",
    "wordnet",
    "omw-1.4",
    "averaged_perceptron_tagger",
    "averaged_perceptron_tagger_eng",
]

for resource in resources:
    nltk.download(resource, quiet=True)

print("NLTK resources ready.")
```

```python
text = (
    "Natural language processing helps computers understand human language. "
    "It transforms raw text into useful representations for analysis. "
    "Good preprocessing improves language tasks such as classification and search."
)

print(text)
```

```python
tokens = word_tokenize(text)
lower_tokens = [token.lower() for token in tokens if token.isalpha()]
stop_words = set(stopwords.words("english"))
content_words = [token for token in lower_tokens if token not in stop_words]

print("Tokens:")
print(tokens)
print()
print("Normalized content words:")
print(content_words)
```

```python
frequency = Counter(content_words)
print("Most common content words:")
for word, count in frequency.most_common():
    print(f"{word}: {count}")
```

```python
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith("J"):
        return wordnet.ADJ
    if treebank_tag.startswith("V"):
        return wordnet.VERB
    if treebank_tag.startswith("N"):
        return wordnet.NOUN
    if treebank_tag.startswith("R"):
        return wordnet.ADV
    return wordnet.NOUN

stem_lemma_rows = []
for token in content_words:
    stemmed = stemmer.stem(token)
    pos = pos_tag([token])[0][1]
    lemmatized = lemmatizer.lemmatize(token, get_wordnet_pos(pos))
    stem_lemma_rows.append((token, stemmed, lemmatized, pos))

print("Token | Stem | Lemma | POS")
print("-" * 40)
for token, stemmed, lemmatized, pos in stem_lemma_rows:
    print(f"{token:12} {stemmed:12} {lemmatized:12} {pos}")
```

```python
tagged_tokens = pos_tag(tokens)
print("POS tags for original tokens:")
for token, tag in tagged_tokens:
    print(f"{token:12} {tag}")
```
