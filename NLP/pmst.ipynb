```python
import re

corpus = """
Hello, my name is Om Ji, withuid 23BAD10012 and official email as 23bad10012@cuchd.in and my whatsapp number is +91 98765 43210. I am skilled in Python, NLP, and SQL.
"""
# Student 24CSE12345 is Priya Sharma. Email: priya.sharma@example.com; phone: +91 98765 43210. Skills: Python, NLP, SQL.
# Student 23AIM67890 is Arjun Mehta. Email: arjun.mehta@university.edu; phone: 987-654-3210. Skills: Java, Machine Learning, Git.
# Student 25ECE54321 is Neha Verma. Email: neha.verma@college.org; phone: (555) 123-4567. Skills: Python, Deep Learning, TensorFlow.
# """

print(corpus)
```

```python
university_id_pattern = re.compile(r"\b\d{2}[A-Za-z]{3}\d{5}\b")
name_pattern = re.compile(r"\b[A-Z][a-z]+(?:\s+[A-Z][a-z]+)+\b")
email_pattern = re.compile(r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b")
phone_pattern = re.compile(r"(?<!\w)(?:\+91[-\s]?\d{5}[-\s]?\d{5}|(?:\(\d{3}\)|\d{3})[-\s]\d{3}[-\s]\d{4})(?!\w)")
skill_pattern = re.compile(
    r"\b(?:Python|NLP|SQL|Java|Machine Learning|Git|Deep Learning|TensorFlow)\b",
    flags=re.IGNORECASE,
)

patterns = {
    "university_ids": university_id_pattern,
    "names": name_pattern,
    "email_ids": email_pattern,
    "phone_numbers": phone_pattern,
    "skills": skill_pattern,
}

patterns
```

```python
extracted = {
    field: pattern.findall(corpus)
    for field, pattern in patterns.items()
}

for field, matches in extracted.items():
    print(f"{field.replace('_', ' ').title()}:")
    for match in matches:
        print(f"  - {match}")
```

```python
records = []
for university_id in university_id_pattern.findall(corpus):
    start = corpus.index(university_id)
    record_text = corpus[start:corpus.find("\n", start)]
    records.append({
        "university_id": university_id,
        "name": name_pattern.search(record_text).group(),
        "email_id": email_pattern.search(record_text).group(),
        "phone_number": phone_pattern.search(record_text).group(),
        "skills": skill_pattern.findall(record_text),
    })

for record in records:
    print(record)
```

```python
assert all(re.fullmatch(r"\d{2}[A-Za-z]{3}\d{5}", item["university_id"]) for item in records)
assert len(records) == 3
assert all(item["email_id"] and item["phone_number"] and item["skills"] for item in records)
print("All requested fields were extracted successfully.")
```
