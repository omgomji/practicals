## Theory

Word generation converts an abstract lexical representation into an appropriate surface form. A generator commonly receives a lemma together with grammatical information such as tense, number, or comparison degree and produces a word such as play → played, child → children, or good → better.

In NLP systems, generation is useful in dialogue systems, machine translation, grammar checking, data augmentation, and natural language generation. Rule-based generators work well for predictable patterns but require exception handling. Lexicon-supported tools combine morphological rules with stored irregular forms. Penn Treebank tags such as VBD, VBG, NNS, JJR, and JJS can be used as compact grammatical specifications for inflection generation.

```python
# Install and import the required library.
from lemminflect import getInflection, getAllInflections
import pandas as pd
import nltk
from nltk import pos_tag, word_tokenize
```

## Penn Treebank Tag Quick Reference

The code below uses Penn Treebank tags as generation constraints. Common examples include:

- VBD: past tense verb
- VBG: gerund or present participle
- NNS: plural noun
- JJR: comparative adjective
- JJS: superlative adjective

```python
# Define lemma and tag pairs for the experiment.
pairs = [
    ("play", "VBD"),
    ("study", "VBG"),
    ("go", "VBD"),
    ("child", "NNS"),
    ("city", "NNS"),
    ("good", "JJR"),
    ("good", "JJS"),
]

records = []
for lemma, tag in pairs:
    forms = getInflection(lemma, tag) or tuple()
    records.append({
        "Lemma": lemma,
        "Tag": tag,
        "Generated Forms": forms,
        "Count": len(forms),
    })

df = pd.DataFrame(records)
df
```

```python
# Present the same results in a clean table for reporting.
report_df = df.copy()
report_df["Generated Forms"] = report_df["Generated Forms"].astype(str)
report_df
```

## Regular and Irregular Patterns

The following examples compare a regular pattern such as play → played with irregular forms such as go → went and child → children.

```python
comparison_examples = [
    ("play", "VBD"),
    ("go", "VBD"),
    ("child", "NNS"),
    ("city", "NNS"),
]

comparison_rows = []
for lemma, tag in comparison_examples:
    forms = getInflection(lemma, tag) or tuple()
    comparison_rows.append({
        "Lemma": lemma,
        "Tag": tag,
        "Generated Form": forms[0] if forms else "",
    })

pd.DataFrame(comparison_rows)
```

## Multiple Valid Forms

Some lemmas can produce more than one valid inflected form for the same tag, especially when both regular and irregular spellings are acceptable.

```python
multi_form_examples = [
    ("dream", "VBD"),
    ("fit", "VBD"),
    ("learn", "VBD"),
]

multi_rows = []
for lemma, tag in multi_form_examples:
    forms = getInflection(lemma, tag) or tuple()
    multi_rows.append({
        "Lemma": lemma,
        "Tag": tag,
        "Generated Forms": forms,
        "Multiple?": len(forms) > 1,
    })

pd.DataFrame(multi_rows)
```

## Optional NLTK Demo

This small demonstration shows how a token sequence can be tagged with Penn Treebank-style POS tags before generation tasks are applied.

```python
sample_text = "Children are playing in the city"
tokens = word_tokenize(sample_text)
tagged = pos_tag(tokens)
tagged
```

## Results

Representative output expected from the main table includes:

- play → VBD → played
- study → VBG → studying
- go → VBD → went
- child → NNS → children
- city → NNS → cities
- good → JJR → better
- good → JJS → best

The experiment demonstrates that lexical knowledge is required for irregular forms, while productive morphological rules handle many regular forms.
