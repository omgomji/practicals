## Theory

Morphology studies the internal structure of words and the rules by which morphemes combine. A **morpheme** is the smallest meaningful grammatical unit. A word may have a root or stem together with affixes. **Inflectional affixes** express grammatical information, such as number or tense, whereas **derivational affixes** create a new lexeme or may change the word class.

Computational morphology supports normalization, information retrieval, machine translation, parsing, and generation. **Stemming** is a surface-oriented normalization method that removes or transforms endings using rules. **Lemmatization** uses vocabulary and grammatical knowledge to return a canonical dictionary form. A simple affix analyzer illustrates segmentation, but robust morphological analysis needs language-specific lexicons and rules.

## Methodology

1. Define a list of words containing prefixes, suffixes, and inflections.
2. Create simple prefix and suffix inventories for demonstration.
3. Detect possible affixes and estimate the remaining stem.
4. Apply Porter and Snowball stemmers.
5. Apply WordNet lemmatization using suitable parts of speech.
6. Compare morphological segments, stems, and lemmas.

## Dataset
The experiment uses a small mixed list of English words: `unhappy`, `replayed`, `kindness`, `studies`, `running`, `nationalization`, `children`, and `disagreement`. It includes derivation, inflection, spelling changes, and an irregular plural.

## Code

```python
import nltk
import pandas as pd
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer

# Download only the lexical resources required by WordNetLemmatizer.
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

porter = PorterStemmer()
snowball = SnowballStemmer('english')
lemmatizer = WordNetLemmatizer()
```

```python
words = ['unhappy', 'replayed', 'kindness', 'studies', 'running',
         'nationalization', 'children', 'disagreement']

# These inventories are intentionally small and are only for demonstration.
prefixes = ['un', 'dis', 're', 'mis', 'pre', 'non']
suffixes = ['ization', 'ness', 'ment', 'ing', 'ied', 'ies', 'ed', 'ly', 's']

# Longest matches are considered first; this avoids matching 's' before 'ness'.
prefixes = sorted(prefixes, key=len, reverse=True)
suffixes = sorted(suffixes, key=len, reverse=True)

def analyze_affixes(word):
    """Return one possible prefix-stem-suffix analysis of a word."""
    prefix = next((p for p in prefixes if word.startswith(p) and len(word) > len(p) + 1), '')
    remaining = word[len(prefix):] if prefix else word
    suffix = next((s for s in suffixes
                   if remaining.endswith(s) and len(remaining) > len(s) + 1), '')
    stem = remaining[:-len(suffix)] if suffix else remaining
    return prefix or '—', stem or '—', suffix or '—'

for word in words:
    prefix, stem, suffix = analyze_affixes(word)
    print(f'{word:18} → prefix: {prefix:4} | stem: {stem:10} | suffix: {suffix}')
```

```python
# Supply a part of speech where it is useful: v = verb, n = noun, a = adjective.
wordnet_pos = {
    'unhappy': 'a', 'replayed': 'v', 'kindness': 'n', 'studies': 'n',
    'running': 'v', 'nationalization': 'n', 'children': 'n', 'disagreement': 'n'
}

rows = []
for word in words:
    prefix, stem, suffix = analyze_affixes(word)
    rows.append({
        'Word': word,
        'Prefix': prefix,
        'Estimated stem': stem,
        'Suffix': suffix,
        'Porter stem': porter.stem(word),
        'Snowball stem': snowball.stem(word),
        'WordNet lemma': lemmatizer.lemmatize(word, pos=wordnet_pos[word])
    })

results = pd.DataFrame(rows)
results
```

## Results and Discussion

The table displays one possible prefix–stem–suffix decomposition beside two stemmer outputs and a WordNet lemma. The affix analyzer correctly demonstrates straightforward cases such as `un + happy`, `re + play + ed`, and `kind + ness`.

Its limitations are visible in forms such as `studies` and `running`: removing a suffix alone produces `stud` or `runn`, while the linguistic base forms are *study* and *run*. The irregular plural `children` has no regular suffix in the simple inventory, but WordNet lemmatization returns *child*. Porter and Snowball may return non-dictionary stems because their purpose is conflation for retrieval rather than dictionary-form recovery.
